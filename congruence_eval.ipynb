```python
import os

def get_image_files(img_dir):
    image_files = []
    for subdir, _, files in os.walk(img_dir):
        for file in files:
            if file.endswith(('.png', '.jpg', '.jpeg')):
                # Get the absolute path and add it to the list
                image_files.append(os.path.abspath(os.path.join(subdir, file)))
    return image_files
```

```python
# Compute FID and KID scores
import clip
from cleanfid import fid

real_mild_images = get_image_files(real_img_dir_1)
real_severe_images = get_image_files(real_img_dir_2)
sd21_i2i_mild_images = get_image_files(sd21_i2i_mild_dir)
sd21_i2i_severe_images = get_image_files(sd21_i2i_severe_dir)
sd21_t2i_mild_images = get_image_files(sd21_t2i_mild_dir)
sd21_t2i_severe_images = get_image_files(sd21_t2i_severe_dir)
sd35_i2i_mild_images = get_image_files(sd35_i2i_mild_dir)
sd35_i2i_severe_images = get_image_files(sd35_i2i_severe_dir)
sd35_t2i_mild_images = get_image_files(sd35_t2i_mild_dir)
sd35_t2i_severe_images = get_image_files(sd35_t2i_severe_dir)
stylegan2_mild_images = get_image_files(stylegan2_mild_dir)
stylegan2_severe_images = get_image_files(stylegan2_severe_dir)
stylegan3_mild_images = get_image_files(stylegan3_mild_dir)
stylegan3_severe_images = get_image_files(stylegan3_severe_dir)



def random_sample(image_set, sample_size=200):
    """
    Randomly sample a specified number of images from a list.
    """

    images = image_set.copy()
    if len(images) < sample_size:
        raise ValueError(f"Not enough images to sample {sample_size}. Available: {len(images)}")
    
    # Ensure reproducibility
    random.seed(42)
    random.shuffle(images)
    partitions = [images[i*200:(i+1)*200] for i in range(5)]

    # Example: access the first partition
    partition1 = partitions[0]
    partition2 = partitions[1]
    partition3 = partitions[2]
    partition4 = partitions[3]
    partition5 = partitions[4]
    return partition1, partition2, partition3, partition4, partition5

```

```python
syn_images = {'sd21_i2i_mild':sd21_i2i_mild_images, 'sd21_i2i_severe':sd21_i2i_severe_images,
             'sd21_t2i_mild':sd21_t2i_mild_images, 'sd21_t2i_severe':sd21_t2i_severe_images,
             'sd35_i2i_mild':sd35_i2i_mild_images, 'sd35_i2i_severe':sd35_i2i_severe_images,
             'sd35_t2i_mild':sd35_t2i_mild_images, 'sd35_t2i_severe':sd35_t2i_severe_images,
             'stylegan2_mild':stylegan2_mild_images, 'stylegan2_severe':stylegan2_severe_images,
             'stylegan3_mild':stylegan3_mild_images, 'stylegan3_severe':stylegan3_severe_images}
real_images = {'mild': real_img_dir_1, 'severe': real_img_dir_2}
```

```python
import tempfile
import shutil
import random
import pandas as pd

fid_score_dict = {}

for key, images in syn_images.items():
    print(f"======== processing {key} images... ========")
    img_type = key.split('_')[-1]  # 'mild' or 'severe'
    if img_type == 'mild':
        real_dir = real_images['mild']
        print(f"======== Using real mild images from {real_dir} for FID computation ========")
    else:
        real_dir = real_images['severe']
        print(f"======== Using real severe images from {real_dir} for FID computation ========")
    sub_dir1, sub_dir2, sub_dir3, sub_dir4, sub_dir5 = random_sample(images)
    temp_dirs = []
    partitions = [sub_dir1, sub_dir2, sub_dir3, sub_dir4, sub_dir5]


    print("======== Creating temporary directories for FID computation ========")
    for idx, partition in enumerate(partitions, 1):
        temp_dir = tempfile.mkdtemp(prefix=f'part{idx}_')
        temp_dirs.append(temp_dir)
        for img_path in partition:
            shutil.copy(img_path, temp_dir)
    
    fid_scores = []
    for temp_dir in temp_dirs:
        fdir1 = temp_dir
        fdir2 = real_dir  # Compare with real mild images
        print(f"Computing FID for {fdir1} against {fdir2}")
        score = fid.compute_fid(fdir1, fdir2, mode="clean", model_name="clip_vit_b_32")
        fid_scores.append(score)
    fid_score_dict[key] = fid_scores
    # Clean up temporary directories
    for temp_dir in temp_dirs:
        shutil.rmtree(temp_dir)

# Convert the dictionary to a DataFrame
fid_df = pd.DataFrame(dict([(k, pd.Series(v)) for k, v in fid_score_dict.items()]))
fid_df = fid_df.map(lambda x: x[0] if isinstance(x, list) and len(x) == 1 else x)
fid_df.to_csv('/data/liangz2/image_quality/fid_scores.csv', index=False)
print("FID scores computed and saved to /data/liangz2/image_quality/fid_scores.csv")
# Print the FID scores
print("FID scores:")
print(fid_df)
    
```

```python
from scipy.stats import ttest_ind
import pandas as pd

# Perform pairwise t-tests between all columns in fid_df
columns = fid_df.columns
ttest_results = {}

for i in range(len(columns)):
    for j in range(i + 1, len(columns)):
        col1, col2 = columns[i], columns[j]
        stat, pval = ttest_ind(fid_df[col1].dropna(), fid_df[col2].dropna(), equal_var=False)
        ttest_results[f"{col1} vs {col2}"] = {'t-statistic': stat, 'p-value': pval}

# Print results
for pair, result in ttest_results.items():
    print(f"{pair}: t-statistic = {result['t-statistic']:.4f}, p-value = {result['p-value']:.4e}")

# Convert t-test results to a DataFrame
ttest_df = pd.DataFrame.from_dict(ttest_results, orient='index')
ttest_df.to_csv('/data/liangz2/image_quality/fid_ttest_results.csv')
print("T-test results saved to /data/liangz2/image_quality/fid_ttest_results.csv")
```

```python
fid_means = fid_df.mean()
fid_stds = fid_df.std()

print("FID Mean scores:\n", fid_means)
print("\nFID Standard deviation:\n", fid_stds)
# Save mean and std in one CSV file with three columns: 'type', 'mean', 'std'
fid_summary_df = pd.DataFrame({
    'type': fid_means.index,
    'mean': fid_means.values,
    'std': fid_stds.values
})
fid_summary_df.to_csv('/data/liangz2/image_quality/fid_mean_std_scores.csv', index=False)
```
